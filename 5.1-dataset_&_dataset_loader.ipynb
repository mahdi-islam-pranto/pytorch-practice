Make a Demo dataset

```python
from sklearn.datasets import make_classification
import torch
```

```python
# Create a synthetic classification dataset using sklearn
X, y = make_classification(
    n_samples=10,       # Number of samples
    n_features=2,       # Number of features
    n_informative=2,    # Number of informative features
    n_redundant=0,      # Number of redundant features
    n_classes=2,        # Number of classes
    random_state=42     # For reproducibility
)
```

```python
print("Input features shape: ", X.shape)
print("Target labels shape: ", y.shape)
```

```python
# convert the data to pytorch tensors
X_tensor = torch.tensor(X, dtype=torch.float32)
y_tensor = torch.tensor(y, dtype=torch.long)
print(X_tensor)
print(y_tensor)
```

Create Custom Dataset for Pytorch

```python
from torch.utils.data import Dataset, DataLoader
```

```python
# dataset class
class CustomDataset(Dataset):
    def __init__ (self, features, labels):
        self.features = features
        self.labels = labels

    def __len__(self):
        return self.features.shape[0]

    def __getitem__(self, index):
        return self.features[index], self.labels[index]
```

```python
# create an instance of the dataset class
dataset = CustomDataset(X_tensor, y_tensor)

print("Length of the dataset: ", len(dataset))
print("2nd sample from the dataset: ", "Feature: ",dataset.features[1], "Label: ", dataset.labels[1])
```

```python
# create data loader
data_loader = DataLoader(dataset, batch_size=2, shuffle=False)
```

```python
# see the full batches of data

for batch_features, batch_labels in data_loader:

  print(batch_features)
  print(batch_labels)
  print("-"*50)
```
