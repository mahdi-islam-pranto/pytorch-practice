Numpy arrays to PyTorch tensors

```python
X_train_tensor = torch.from_numpy(X_train.astype(np.float32))
X_test_tensor = torch.from_numpy(X_test.astype(np.float32))
y_train_tensor = torch.from_numpy(y_train.astype(np.float32))
y_test_tensor = torch.from_numpy(y_test.astype(np.float32))
print(f"X train: {X_train_tensor.shape}")
print(f"y train: {y_train_tensor.shape}")
```

Defining the pytorch Model

# Neural Network Architecture: `MySimpleNN`

This model is a simple feed-forward neural network designed for binary classification on the breast cancer dataset.

## Architecture Overview

Input: 30 features
- Hidden Layer 1: Linear(30 -> 16)
- Activation: ReLU
- Output Layer: Linear(16 -> 1)
- Activation: Sigmoid

## Layer-by-Layer Description

### 1) Input Layer
- Input dimension: 30
- This corresponds to the 30 standardized feature columns used in `X_train_tensor`

### 2) Hidden Layer 1
`nn.Linear(in_features=30, out_features=16)`

- Weight matrix shape: (16, 30)
- Number of weights: 16 × 30 = 480
- Bias vector shape: (16,)
- Number of biases: 16
- Trainable parameters: 480 + 16 = 496

Activation:
`nn.ReLU()`

- No trainable parameters

### 3) Output Layer
`nn.Linear(in_features=16, out_features=1)`

- Weight matrix shape: (1, 16)
- Number of weights: 1 × 16 = 16
- Bias vector shape: (1,)
- Number of biases: 1
- Trainable parameters: 16 + 1 = 17

Activation:
`nn.Sigmoid()`

- No trainable parameters

## Total Number of Parameters

Total trainable parameters = 496 + 17 = 513

## Model Summary

- Input features: 30
- Hidden units: 16
- Output units: 1
- Activation functions: ReLU and Sigmoid
- Loss function: `BCELoss()`
- Optimizer: `SGD`
- Learning rate: 0.1
- Total parameters: 513

## Mathematical Representation

For an input vector x ∈ R^30:

1. Hidden layer:
    z1 = W1 x + b1
    a1 = ReLU(z1)

2. Output layer:
    z2 = W2 a1 + b2
    ŷ = Sigmoid(z2)

where:
- W1 has shape (16, 30)
- b1 has shape (16,)
- W2 has shape (1, 16)
- b2 has shape (1,)

This network is intentionally small and lightweight, making it suitable for binary classification tasks such as benign vs malignant tumor detection.

```python
import torch.nn as nn

class MySimpleNN(nn.Module):

  def __init__(self, num_features):

    super().__init__()
    # define the layers of the model
    self.layers = nn.Sequential(
      nn.Linear(in_features=num_features, out_features=16),
      nn.ReLU(),
      nn.Linear(in_features=16, out_features=1),
      nn.Sigmoid()
    )

  # forward method is where the data passes through the layers of the model
  def forward(self, features):

    return self.layers(features)
```

```python
# parameters
learning_rate = 0.1
epochs = 25
```

```python
# define loss function
loss_function = nn.BCELoss()
```

Training Pipeline

```python
# create model
model = MySimpleNN(X_train_tensor.shape[1])

# define optimizer
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

# define loop
for epoch in range(epochs):

  # forward pass
  y_pred = model(X_train_tensor)

  # loss calculate
  loss = loss_function(y_pred, y_train_tensor.view(-1,1))

  # clear gradients
  optimizer.zero_grad()

  # backward pass
  loss.backward()

  # parameters update
  optimizer.step()

  # print loss in each epoch
  print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```

Evaluation

```python
# model evaluation
with torch.no_grad():
  y_pred = model(X_test_tensor)
  y_pred = (y_pred > 0.5).float()
  accuracy = (y_pred == y_test_tensor).float().mean()
  print(f'Accuracy: {accuracy.item()}')
```
