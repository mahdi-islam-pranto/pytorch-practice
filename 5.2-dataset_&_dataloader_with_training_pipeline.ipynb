Prepare the Dataset

```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace= True)
df.head()
```

```python
# train test split
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)
```

```python
# data scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
# convert the numpy arrays to tensors
X_train_tensor = torch.from_numpy(X_train.astype(np.float32))
X_test_tensor = torch.from_numpy(X_test.astype(np.float32))
y_train_tensor = torch.from_numpy(y_train.astype(np.float32))
y_test_tensor = torch.from_numpy(y_test.astype(np.float32))
```

```python
print("Features shape: ", X_train_tensor.shape)
print("Label shape: ", y_train_tensor.shape)
```

Make Custom Dataset Class for Pytorch

```python
from torch.utils.data import Dataset, DataLoader
```

```python
class CustomDataset(Dataset):
    def __init__ (self, features, labels):
        self.features = features
        self.labels = labels
        
    def __len__(self):
        return self.features.shape[0]
        
    def __getitem__(self, index):
        return self.features[index], self.labels[index]
```

```python
# train dataset and test dataset instance
train_dataset = CustomDataset(X_train_tensor, y_train_tensor)
test_dataset = CustomDataset(X_test_tensor, y_test_tensor)
```

```python
train_dataset[10]
```

Create Data Loader

```python
# create data loader for both train and test dataset
train_loader = DataLoader(dataset=train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(dataset=test_dataset, batch_size=32, shuffle=True)
```

Defining The Model

```python
import torch.nn as nn
```

```python
# create the model class
class BreastCancerModel(nn.Module):
    def __init__(self, num_features):
        super().__init__()
        # define the layers of the model
        self.layers = nn.Sequential(
            nn.Linear(num_features, 16),
            nn.ReLU(),
            nn.Linear(16, 8),
            nn.ReLU(),
            nn.Linear(8, 1),
            nn.Sigmoid()
        )

    def forward(self, features):
        return self.layers(features)
```

```python
# parameters
learning_rate = 0.1
epochs = 25
```

```python
# define the model
model = BreastCancerModel(X_train_tensor.shape[1])

# define the loss function
loss_fn = nn.BCELoss()

# define the optimizer
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)
```

Training Pipeline

```python
# define the training loop
def train_model(model, train_loader, loss_fn, optimizer, epochs):
    for epoch in range(epochs):
        for batch_features, batch_labels in train_loader:
            # forward pass
            y_pred = model(batch_features)

            # compute loss
            loss = loss_fn(y_pred, batch_labels.view(-1,1))

            # clear gradients
            optimizer.zero_grad()

            # backward pass
            loss.backward()

            # update parameters
            optimizer.step()

        # print loss in each epoch
        print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```

```python
# train the model
train_model(model, train_loader, loss_fn, optimizer, epochs)
```

```python
# Model evaluation using test_loader
model.eval()  # Set the model to evaluation mode
accuracy_list = []

with torch.no_grad():
    for batch_features, batch_labels in test_loader:
        # Forward pass
        y_pred = model(batch_features)
        y_pred = (y_pred > 0.8).float()  # Convert probabilities to binary predictions

        # Calculate accuracy for the current batch
        batch_accuracy = (y_pred.view(-1) == batch_labels).float().mean().item()
        accuracy_list.append(batch_accuracy)

# Calculate overall accuracy
overall_accuracy = sum(accuracy_list) / len(accuracy_list)
print(f'Accuracy: {overall_accuracy:.4f}')
```

```python
# predicting on a single sample from the test set
model.eval()
with torch.no_grad():
    sample_index = 0  # Change this index to test different samples
    sample_features, sample_label = test_dataset[sample_index]
    sample_features = sample_features.unsqueeze(0)  # Add batch dimension

    prediction = model(sample_features)
    predicted_class = (prediction > 0.8).float().item()  # Convert probability to binary prediction

    print(f'Sample Features: {sample_features}')
    print(f'True Label: {sample_label.item()}')
    print(f'Predicted Class: {predicted_class}')
```
